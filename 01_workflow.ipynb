## What happens
1. Validate the source package; extract terms using saved answers or an explicitly started Nebius request.
2. Live output pauses for source-bound human fact verification before financial calculations.
3. Python computes financial metrics, synthetic policy checks and illustrative risk scores.
4. Synthesis proposes an AI outcome; live conclusions require separate human evidence verification.
5. Deterministic readiness prevents an AI Approve proposal or Low score from overriding mandatory blockers. An authorized credit decision is a separate action where routed.
6. The report records findings, readiness, evidence and human decision status. Pending Review is not automatically Reject.
